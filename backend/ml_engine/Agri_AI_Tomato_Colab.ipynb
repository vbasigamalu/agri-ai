### Step 3: Upload and Extract Dataset
Drag & drop your `tomato_dataset.zip` file into the left file sidebar in Colab, or run the cell below to upload it.

```python
import os, zipfile
from google.colab import files

if not os.path.exists('tomato_dataset.zip') and not os.path.exists('dataset'):
    print('Please upload tomato_dataset.zip:')
    uploaded = files.upload()

if os.path.exists('tomato_dataset.zip') and not os.path.exists('dataset'):
    print('Extracting dataset...')
    with zipfile.ZipFile('tomato_dataset.zip', 'r') as z:
        z.extractall('.')
    print('✅ Extracted successfully!')
```
